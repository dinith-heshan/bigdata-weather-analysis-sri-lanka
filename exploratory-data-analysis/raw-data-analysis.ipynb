```python
import pandas as pd
```

```python
weather_data_raw_df = pd.read_csv(r'../data/raw/weatherData.csv')
```

```python
weather_data_raw_df
```

```python
weather_data_date_parsed_df = weather_data_raw_df.copy()
weather_data_date_parsed_df['date'] = pd.to_datetime(weather_data_raw_df['date'], format='%m/%d/%Y')
```

```python
weather_data_may_df = weather_data_date_parsed_df[weather_data_date_parsed_df['date'].dt.month == 5]
```

```python
weather_data_may_df
```

```python
weather_data_selected_df = weather_data_may_df[['sunshine_duration (s)', 'precipitation_hours (h)', 'wind_speed_10m_max (km/h)', 'et0_fao_evapotranspiration (mm)']]
```

```python
weather_data_selected_df
```

```python
weather_data_unique_rows_df = weather_data_selected_df[~weather_data_selected_df.duplicated()]
```

```python
weather_data_unique_rows_df
```

```python
weather_data_unique_features_df = (
    weather_data_unique_rows_df
    .groupby(
        [
            'sunshine_duration (s)',
            'precipitation_hours (h)',
            'wind_speed_10m_max (km/h)'
        ],
        as_index=False
    )
    .agg(
        et0_mean=('et0_fao_evapotranspiration (mm)', 'mean'),
        et0_std=('et0_fao_evapotranspiration (mm)', 'std'),
        sample_count=('et0_fao_evapotranspiration (mm)', 'count')
    )
)
```
